# 07 — Post-hoc hedge (NOT pre-registered): provider-selected decode of the same ensemble

The adopted candidate (`exp3_ref_ens3`, notebook 05) is decoded for `L_ctrl`. Because BAR-ATH's ground-truth definition is unknown (REPORT §7), this notebook writes a second file from the *same* cached exp_2 ensemble, decoded with the config selected on the 7 `target_like` out-of-fold predictions against **provider** labels. It does not change the pre-registered candidate. Aggregates only; validator; byte-identical rebuild from 3 independent fits; no public check.

```python
from pathlib import Path
import os, sys, json, time
ROOT = Path.cwd()
if not (ROOT / 'exp3_common.py').exists():
    raise RuntimeError('Run with working directory exp_jeremy/exp_3')
sys.path.insert(0, str(ROOT))
sys.dont_write_bytecode = True
import numpy as np
import pandas as pd
pd.set_option('display.width', 220)
pd.set_option('display.max_columns', 40)
import exp3_common as C
import exp1_common as C1
protocol = C.verify_prereg()
print('PREREG verified', protocol['prereg_sha256'][:16], '| frozen', protocol['frozen_at_utc'])
import exp3_target as E
assert C.public_check_spent(), 'exp_1 public check marker missing'
```

```python
t0 = time.time()
match, items, fusion = E.inputs()
models = {arm: E.full_model(arm) for arm in E.MEMBERS[E.HEDGE_VARIANT]}
summary = E.build_hedge(match, items, models, fusion)
summary['reproduction'] = E.reproduce_hedge(match, items, fusion)
C1.write_json(C.RESULTS / 'target_hedge_provider.json', summary)
print(json.dumps({k: v for k, v in summary.items() if k != 'share_graded'}, indent=1))
print(f'{time.time() - t0:.0f}s')
```
