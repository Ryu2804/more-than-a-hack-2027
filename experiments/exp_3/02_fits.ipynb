# 02 — `exp3_ctrl` fits (3 seeds × 28 nested fits)

exp2_aug recipe (round 0 original + rounds 1–2 `target_aug`, mirror, `TrainConfig()`) with `L_ctrl` training targets. Each fit predicts both views of its held-out matches and saves its weights. Cached by fingerprint (includes the `L_ctrl` code hash); progress in `results/fit_log.txt`.

```python
from pathlib import Path
import os, sys, json, time
ROOT = Path.cwd()
if not (ROOT / 'exp3_common.py').exists():
    raise RuntimeError('Run with working directory exp_jeremy/exp_3')
sys.path.insert(0, str(ROOT))
sys.dont_write_bytecode = True
import numpy as np
import pandas as pd
pd.set_option('display.width', 220)
pd.set_option('display.max_columns', 40)
import exp3_common as C
import exp1_common as C1
protocol = C.verify_prereg()
print('PREREG verified', protocol['prereg_sha256'][:16], '| frozen', protocol['frozen_at_utc'])
import exp3_protocol as P, exp1_net as N
print('device', N.DEVICE, '| config', N.TrainConfig().as_dict())
```

```python
log_path = C.RESULTS / 'fit_log.txt'
def log(message, **_):
    print(message, flush=True)
    with log_path.open('a') as handle:
        handle.write(message + '\n')
t0 = time.time()
for arm in P.MODEL_ARMS:
    start = time.time()
    P.ensure_predictions(arm, log=log)
    log(f'ARM DONE {arm} {time.time() - start:.0f}s')
log(f'ALL FITS DONE {time.time() - t0:.0f}s, exp_3 fits cached {P.exp3_fit_count()}')
```
